```python
import sys
sys.path.append(r"c:\Users\pablo\Downloads\MLFromScratch")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
from sklearn.model_selection import train_test_split
from ML.LogisticRegression import LogisticRegression
from sklearn.linear_model import LogisticRegression as SklearnLogisticRegression
```

```python
# Load and clean the breast cancer classification dataset
data = pd.read_csv(r"C:\Users\pablo\Downloads\MLFromScratch\datasets\CancerClassification.csv")
data.to_numpy()  # Convert DataFrame to NumPy array to make it compatible with the library's implementation
X = data.drop(columns=['id', 'diagnosis', 'Unnamed: 32']).values
y = data['diagnosis'].map({'M': 1, 'B': 0}).values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
MLFromScratchLR = LogisticRegression()
MLFromScratchLR.fit(X_train, y_train)
MLFromScratch_predictions = MLFromScratchLR.predict(X_test)
MLFromScratch_accuracy = np.mean(MLFromScratch_predictions == y_test) * 100
print(f"MLFromScratch Accuracy: {MLFromScratch_accuracy:.2f}%")

ScikitLearnLR = SklearnLogisticRegression()
ScikitLearnLR.fit(X_train, y_train)
ScikitLearn_predictions = ScikitLearnLR.predict(X_test)
ScikitLearn_accuracy = np.mean(ScikitLearn_predictions == y_test) * 100
print(f"Scikit-learn Accuracy: {ScikitLearn_accuracy:.2f}%")
```
