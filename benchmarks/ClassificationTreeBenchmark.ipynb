```python
import sys
sys.path.append(r"c:\Users\pablo\Downloads\MLFromScratch")
import numpy as np
import pandas as pd
from ML.base import BaseModel
from collections import Counter
from ML.ClassificationTree import ClassificationTree
from sklearn.tree import DecisionTreeClassifier as SklearnClassificationTree
from sklearn.model_selection import train_test_split
from abc import ABC, abstractmethod
```

```python
# Load and clean the breast cancer classification dataset
data = pd.read_csv(r"C:\Users\pablo\Downloads\MLFromScratch\datasets\CancerClassification.csv")

X = data.drop(columns=['id', 'diagnosis', 'Unnamed: 32']).values
y = data['diagnosis'].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
classification_tree = ClassificationTree(min_samples_split = 3, max_depth = 5)
classification_tree.fit(X_train, y_train)
predictions = classification_tree.predict(X_test)

accuracy = np.sum(predictions == y_test) / len(y_test)
print(f"Accuracy of MLFromScratch Classification Tree: {(accuracy*100):.2f}%")

SKlearn_tree = SklearnClassificationTree(min_samples_split = 3, max_depth = 5)
SKlearn_tree.fit(X_train, y_train)
Sklearn_predictions = SKlearn_tree.predict(X_test)
Sklearn_accuracy = np.sum(Sklearn_predictions == y_test) / len(y_test)
print(f"Accuracy of the sklearn Classification Tree: {(Sklearn_accuracy*100):.2f}%")
```
