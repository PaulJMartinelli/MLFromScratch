```python
import sys
sys.path.append(r"c:\Users\pablo\Downloads\MLFromScratch")
import numpy as np
import pandas as pd
from collections import Counter
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from ML.KNN import KNN
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Load and clean the breast cancer classification dataset
data = pd.read_csv(r"C:\Users\pablo\Downloads\MLFromScratch\datasets\CancerClassification.csv")

X = data.drop(columns=['id', 'diagnosis', 'Unnamed: 32']).values
y = data['diagnosis'].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
plt.scatter(X_train[y_train == 'B', 0], X_train[y_train == 'B', 1], color='tab:green', label='Benign')
plt.scatter(X_train[y_train == 'M', 0], X_train[y_train == 'M', 1], color='tab:red', label='Malignant')
plt.xlabel('Radius Mean')
plt.ylabel('Texture Mean')
plt.legend()
plt.show()
```

```python
knn = KNN(k=7)
knn.fit(X_train, y_train)
predictions = knn.predict(X_test)
accuracy = np.mean(predictions == y_test) * 100
print(f"MLFromScratch Accuracy: {accuracy:.2f}%")
scikitknn = KNeighborsClassifier(n_neighbors=7)
scikitknn.fit(X_train, y_train)
predictions = scikitknn.predict(X_test)
scikit_accuracy = np.mean(predictions == y_test) * 100
print(f"Scikit-learn Accuracy: {scikit_accuracy:.2f}%")
```

```python
# visualization of MLFromScratch predictions
plt.scatter(X_test[predictions == 'B', 0], X_test[predictions == 'B', 1], color='tab:green', label='Predicted Benign', marker = 'x')
plt.scatter(X_test[predictions == 'M', 0], X_test[predictions == 'M', 1], color='tab:red', label='Predicted Malignant', marker = 'x')
plt.xlabel('Radius Mean')
plt.ylabel('Texture Mean')
plt.legend()
plt.show()
```
